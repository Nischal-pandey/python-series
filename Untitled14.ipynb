```python
fruits = ["Mango","Apple","Banana","orange"]
for fruit in fruits:
    print(fruit)
```

```python
numbers = [55,66,88,93,18,63]
total = 0
for num in numbers:
    total += num
print(f"total {total}")
print(f"Average {total/len(numbers)}")
```

```python
students = ["Rahul", "Priya", "Amit","Neha"]

for index, name in enumerate(students):
    print(f"{index}. {name}")
```

```python
# simple string loop
word = "Python"

for char in word:
    print(char,end=" ")
print()
```

```python
# reverse srring
word = "Hello"
reversed_word = ""
for char in word:
    reversed_word = char + reversed_word
print(f"Reversed {reversed_word}")
```

```python
# count voe=wels
text = "Hello World Python"
vowels = 0
consonants = 0
for char in text.lower():
    if char in "aeiou":
        vowels += 1
    elif char.isalpha():
        consonants += 1
print(f"Vowels {vowels}")
print(f"Consonants {consonants}")
```

```python
# Character frequency
word = "mississippi"
freq = {}
for char in word:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1
print(f"Frequency {freq}")
```

```python
# Tuple with loop
colors = ("red", "green", "Blue", "Yellow")
for colors in colors:
    print(f"{colors}")
    
```

```python
# tuple unpacking in loop
students = [
    ("Rahul", 85),
    ("Mohan",70),
    ("Amit",5),
    ("Neha",60)
]

for name, marks in students:
    status = "Pass" if marks >= 33 else "Fail"
    print(f"{name}: {marks} marks -> {status}")
```

```python
# for loop with dictionary
student = {
    "name": "Rahul",
    "age" : 18,
    "marks":85,
    "city" : "Delhi"
}

# Key pe loop (default)
# for key in student:
#     print(f"{key}")

# Values pe loop
# for value in student.values():
#     print(f"{value}")

# keys + values dono pe loop 
for key, value in student.items():
    print(f"{key}:{value}")
```

```python
# for loop with set
# set pe loop (order guarantee nahi)
unique_numbers = {3,1,4,8,9,5,7,5,6}

for num in unique_numbers:
    print(num, end="")
```

```python
# for loop with zip
#ZIP ()= MULTIPLE LISTS KO PARRRALEL LOOP

names = ["Rahul","Priya","Amit"]
marks = [85,92,78]
cities = ["Delhi", "Mumbai", "Pune"]

for name, mark, city in zip(names, marks, cities):
    print(f"{name} {marks} {city}")
```
