```python
# SET OPERATIONS
```

```python
class_A = {"Ram","Shyam","Geeta","Sita","Mohan"}
class_B = {"Geeta","Mohan","Ravi","Priya","Amit"}

print(" Class A students:", class_A)
print(" Class B students:", class_B)
print("\nNotice: Geeta aur Mohan DONO classes mein hain!\n")
```

```python
# UNION - ALL ITEMS FROM BOTH SETS(SABHI ITEMS)
```

```python
all_students = class_A | class_B
print("union using |:")
print(all_students)
```

```python
# Method 2 : using union() operator
all_students = class_A.union(class_B)
print("\nUNION using union:")
print(all_students)
```
